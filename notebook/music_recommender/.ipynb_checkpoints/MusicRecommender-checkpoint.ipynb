```python
songs_data = []

for file in h5_files:
    song = h5py.File(file,'r')
    analysis = song["analysis"]["songs"][0]
    metadata = song["metadata"]["songs"][0]
    artist_terms = song["metadata"]["artist_terms"][:]
    song_info = {
        "title": metadata["title"],
        "artist_name": metadata["artist_name"],
        "tempo": analysis["tempo"],
        "loudness": analysis["loudness"],
        "duration": analysis["duration"],
        "key": analysis["key"],
        "mode": analysis["mode"],
        "time_signature": analysis["time_signature"],
         "artist_terms": artist_terms
    }
    songs_data.append(song_info)
```

```python
len(songs_data)
```

```python
songs = pd.DataFrame(songs_data)
```

```python
songs.head()
```

```python
songs['artist_name']= songs['artist_name'].str.decode("utf-8")
songs['title']= songs['title'].str.decode("utf-8")
songs['artist_terms']= songs['artist_terms'].apply(
    lambda terms: [term.decode('utf-8') for term in terms]
)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
features = songs[['tempo', 'loudness', 'key']]
```

```python
scaler.fit(features)
```

```python
scaled_features = scaler.transform(features)
```

```python
scaled_features
```

```python
scaled_features= pd.DataFrame(scaled_features,columns=['tempo', 'loudness', 'key'])
```

```python
scaled_features
```

```python
recommendation_features = pd.concat(
    [scaled_features, artist_features],
    axis=1
)
```

```python
recommendation_features.shape
```

```python
recommendation_features.head(5)
recommendation_features.iloc[[   0, 3523, 8939, 2121, 1423, 9565]]
```

```python
from sklearn.neighbors import NearestNeighbors
```

```python
knn = NearestNeighbors(n_neighbors=21)
```

```python
knn.fit(recommendation_features)
```

```python
distances, indices = knn.kneighbors(
    recommendation_features.iloc[[0]]
)
```

```python
indices 
```

```python
songs.iloc[indices[0][1:]]
```

```python
matching_song = songs[songs['title'] == "Trash (live Paris)"]

matching_song_index = matching_song.index[0]

distances, indices = knn.kneighbors(
    recommendation_features.iloc[[matching_song_index]]
)

songs.iloc[indices[0][1:]]
```

```python
#fixing the weight of artist features on the dress
```

```python
weighted_artist_features =  artist_features * 0.25 #we want the weight of the artists genre to be less 
```

```python
recommendation_features = pd.concat(
    [scaled_features, weighted_artist_features],
    axis=1
)
```

```python
knn.fit(recommendation_features)
```

```python
distances, indices = knn.kneighbors(
    recommendation_features.iloc[[0]]
)
```

```python
def recommended_songs(song_name):
    #song_name = input('Enter a song name: ')

    matching_song = songs[songs['title'] == song_name]

    if matching_song.empty:
        return "Song not found"

    matching_song_index = matching_song.index[0]

    distances, indices = knn.kneighbors(
        recommendation_features.iloc[[matching_song_index]]
    )

    return songs.iloc[indices[0][1:]]
```

```python
recommended_songs("Breakin' Dishes")
```
